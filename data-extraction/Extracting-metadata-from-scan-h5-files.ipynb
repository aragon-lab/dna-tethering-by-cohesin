```python
import lumicks.pylake as lk
import matplotlib.pyplot as plt
import numpy as np
```

```python
path = "./examples/scan-extraction/input/20230324-145634 Scan 32.h5"
file = lk.File(path)

print('Distance 1 = '+str(np.mean(file.distance1.data)))
print('Force 1x = '+str(np.mean(file.force1x.data)))
print('Force 2x = '+str(np.mean(file.force2x.data)))

plt.figure()
plt.plot(file.force1x.data)
plt.title('Force 1x')
plt.show()

plt.figure()
plt.plot(file.force2x.data)
plt.title('Force 2x')
plt.show()
```
