```python
# Importing libraries
from lumicks import pylake
import os
```

```python
# Setting parameters
root = '/Users/gfisher/Downloads/OneDrive_1_27-10-2025/'

for path in os.listdir(root):
    if (os.path.splitext(path)[1] != ".h5"):
        continue
    
    print("Processing "+root+path)
    out_name, ext = os.path.splitext(root+path)
    
    # Loading data
    file = pylake.File(root+path)

    for idx,kymo in enumerate(file.kymos.values()):   
        if "save_tiff" in dir(kymo):
            kymo.save_tiff("%s_kymograph%i.tif" % (out_name,idx))   
        elif "export_tiff" in dir(kymo):
            kymo.export_tiff("%s_kymograph%i.tif" % (out_name,idx))   
        else:
            print("No save function found in kymo")
```
