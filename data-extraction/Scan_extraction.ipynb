```python
from lumicks import pylake
from kymotools.utils import fileutils

import os
```

```python
#Load data
data_folder = '/Users/gfisher/Downloads/OneDrive_1_27-10-2025 (1)/' #Set as folder containing data to process
output_folder = (data_folder+"output/") #Folder used for outputs
exp_id_split = data_folder.split("/")
exp_id = exp_id_split[-1]

#Create output folder
if not os.path.exists(output_folder + exp_id):
    os.makedirs(output_folder + exp_id)
    
#Get All Files from data folder
filenames = sorted(os.listdir(data_folder)) #Sorted based on file names

#Load all Kymographs, FD curves, .tiff, marker and scan files to lists
for count,filename in enumerate(filenames):
    if filename.endswith(".h5") and "Scan" in filename:
        print("Loading " + str(count+1) + "/" + str(len(filenames)) + ": " + filename)

        file = pylake.File(data_folder + filename)
        for scan_idx in file.scans:
            scan = file.scans[scan_idx]
            scan.export_tiff(output_folder + exp_id + "/" + fileutils.strip_ext(filename) + "_" + scan_idx + ".tiff")
        
```
