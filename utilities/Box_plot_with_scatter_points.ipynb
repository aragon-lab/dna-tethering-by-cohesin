```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from scipy.stats import ttest_ind
```

```python
path = "/Users/sc13967/Library/CloudStorage/OneDrive-UniversityofBristol/People/Gemma Fisher/2026-07-31 Binding retention with HSW/Images/Tiffs/"
filename = "2026-07-31 Retention percentage.csv"

import csv
vals = [[],[],[],[],[],[]]
with open(f"{path}{filename}") as f:
    lines = f.readlines()
    for line in lines:
        line = line.replace("\ufeff","")
        line = line.replace("\n","")
        curr = line.split(",")
        for i in range(6):            
            try:
                vals[i].append(float(curr[i]))
            except:
                continue
```
