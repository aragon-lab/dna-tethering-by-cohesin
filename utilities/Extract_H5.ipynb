```python
from lumicks import pylake

import os
```

```python
input_path = "/Users/sc13967/Library/CloudStorage/OneDrive-UniversityofBristol/People/Gemma Fisher/2026-06-30 Final figures/HFstripped_Raw data for Zenodo_V2/Figure 1/Figure 1E/Kymograph/"
output_path = input_path
```

```python
for filename in os.listdir(input_path):
        if os.path.splitext(filename)[1] != ".h5":
                continue
                
        (rootname, ext) = os.path.splitext(filename)
        
        file = pylake.File(f"{input_path}{filename}")
        
        for (idx, kymo) in enumerate(file.kymos.values()):
                kymo.export_tiff(f"{output_path}{rootname}_kymo{idx}.tif")
                
        for (idx, scan) in enumerate(file.scans.values()):
                scan.export_tiff(f"{output_path}{rootname}_scan{idx}.tif")
```
